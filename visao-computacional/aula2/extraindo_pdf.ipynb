```python
from paddleocr import PaddleOCR

ocr = PaddleOCR(use_angle_cls=True, lang='en')

# Função para extrair texto de um PDF
def extrair_texto_pdf(caminho_pdf):
    # Use o OCR para extrair texto do PDF
    resultado = ocr.ocr(caminho_pdf, cls=True)
    
    # Concatenar o texto extraído
    texto_extraido = ""
    for linha in resultado:
        for palavra in linha:
            texto_extraido += palavra[1][0] + " " # o indice 0 representa o texto da palavra
        texto_extraido += "\n"
    
    return texto_extraido


# Caminho para o arquivo PDF
caminho_pdf = 'ana_curriculo.pdf'

# Extrair texto do PDF
texto = extrair_texto_pdf(caminho_pdf)

# Imprimir o texto extraído
print(texto)
```
