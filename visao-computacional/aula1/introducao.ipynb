```python
import cv2
import matplotlib.pyplot as plt

imagem = cv2.imread('foto.jpg')

# Exibir a imagem usando OpenCV (sem usar Matplotlib)
# cv2.imshow('Imagem', imagem)
# cv2.waitKey(0)
# cv2.destroyAllWindows()

imagem_rgb = cv2.cvtColor(imagem, cv2.COLOR_BGR2RGB)
plt.imshow(imagem_rgb)
plt.axis('off') # Ocultar os eixos
plt.show()

```

```python
# Converter a imagem para escala de cinza
imagem_cinza = cv2.cvtColor(imagem, cv2.COLOR_BGR2GRAY)

# Exibir a imagem em escala de cinza usando Matplotlib
plt.imshow(imagem_cinza, cmap='gray')
plt.axis('off') # Ocultar os eixos
plt.show()
```

```python
#Redimensionar a imagem para um tamanho específico (por exemplo, 100x150 pixels)
imagem_redimensionada = cv2.resize(imagem, (100, 150), interpolation=cv2.INTER_AREA)

# Exibir a imagem redimensionada usando Matplotlib
imagem_redimensionada_rgb = cv2.cvtColor(imagem_redimensionada, cv2.COLOR_BGR2RGB)
plt.imshow(imagem_redimensionada_rgb)
plt.axis('off') # Ocultar os eixos
plt.show()
```

```python
# Suavizar a imagem usando um filtro gaussiano
imagem_suavizada = cv2.GaussianBlur(imagem, (15, 15), 0)

# Exibir a imagem suavizada usando Matplotlib
imagem_suavizada_rgb = cv2.cvtColor(imagem_suavizada, cv2.COLOR_BGR2RGB)
plt.imshow(imagem_suavizada_rgb)
plt.axis('off') # Ocultar os eixos
plt.show()
```

```python
#Desenhar um retângulo azul em torno de um objeto específico na imagem

# Definir as coordenadasea cor do retângulo

inicio = (300, 200)  # deslocamento do ponto superior esquerdo do retângulo, deslocamento de cima para baixo
fim = (520, 650)   # deslocamento da direita para a esquerda do ponto inferior direito do retângulo
cor = (255, 0, 0)  # cor azul em formato BGR
espessura = 2  # espessura da borda do retângulo

# Desenhar o retângulo na imagem
imagem_com_retangulo = cv2.rectangle(imagem.copy(), inicio, fim, cor, espessura)

# Converter a imagem para RGB para exibição com Matplotlib
imagem_com_retangulo_rgb = cv2.cvtColor(imagem_com_retangulo, cv2.COLOR_BGR2RGB)
plt.imshow(imagem_com_retangulo_rgb)
plt.axis('off') # Ocultar os eixos
plt.show()

```
