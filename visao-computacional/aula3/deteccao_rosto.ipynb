```python
import cv2
import matplotlib.pyplot as plt

face_cascade = cv2.CascadeClassifier(cv2.data.haarcascades + 'haarcascade_frontalface_default.xml')

imagem = cv2.imread('rosto.webp')

imagem_cinza = cv2.cvtColor(imagem, cv2.COLOR_BGR2GRAY)

# Detectar rostos na imagem
rostos = face_cascade.detectMultiScale(imagem, scaleFactor=1.1, minNeighbors=5, minSize=(55, 55))

# Desenhar retângulos ao redor dos rostos detectados
for (x, y, w, h) in rostos:
    cv2.rectangle(imagem, (x, y), (x + w, y + h), (255, 0, 0), 2)

# Desenhar retâmgulos ao redor das gaces + cabelos detectados
for (x, y, w, h) in rostos:
    # Definir a região do cabelo e rosto combinados
    regiao_cabelo = max(0, y - int( h / 2))
    combined_height  = h + (y - regiao_cabelo)

    # Desenhar o retângulo ao redor do rosto e cabelo
    cv2.rectangle(imagem, (x, regiao_cabelo), (x + w, regiao_cabelo + combined_height), (255, 0, 0), 2)

imagem = cv2.cvtColor(imagem, cv2.COLOR_BGR2RGB)

plt.imshow(imagem)
plt.axis('off')
plt.show()
```
